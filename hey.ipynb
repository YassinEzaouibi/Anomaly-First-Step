```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Step one: Load the data
df = pd.read_csv('./data/simple_timeseries.csv', parse_dates=['timestamp'])

# Step 2: Plot it
plt.figure(figsize=(10, 4))
plt.plot(df['timestamp'], df['value'], marker='o')
plt.title('Simple Time Series')
plt.xlabel('Timestamp')
plt.ylabel('Value')
plt.grid(True)
plt.show()
```

```python

# Step 1: Calculate mean and std
mean = df['value'].mean()
std = df['value'].std()


# Step 2: Compute Z-scores
df['z_score'] = (df['value'] - mean) / std

# Step 3: Flag anomalies (z > 3 or z < -3)
df['anomaly'] = df['z_score'].abs() > 2

# Step 4: Plot with anomaly markers
plt.figure(figsize=(10, 4))
plt.plot(df['timestamp'], df['value'], label='Value')
plt.scatter(df[df['anomaly']]['timestamp'], df[df['anomaly']]['value'], color='red', label='Anomaly', zorder=5)
plt.title('Anomaly Detection (Z-Score)')
plt.xlabel('Timestamp')
plt.ylabel('Value')
plt.legend()
plt.grid(True)
plt.show()

# Print the anomaly row(s)
print(df[df['anomaly']])
```
